### 2. Model Training

Your exercise is now to train a logistic regression model using an one-vs-rest approach (see lecture slides).

For this, implement the following steps:
1. Apply min-max normalization to `X_train` and `X_test` to normalize the data.
2. Train a binary logistic regression to predict: "Is there the digit `9` on the image ?". Hint: For this you need to change the labels in `y_train`. Calculate the accuaracy of this classifer on the test data (you need to change also the labels in `y_test`).
3. Train a binary logistic regression on each digit from `0` to `9`. Save each model in a list.
4. For every test data point, predict which each model on it to find out which models returns the highest probability for the point. Hint: Use `.predict_proba()` to get class probabilities from each binary classifier.
5. Use the results from step 4 to calculate the accuracy of the overall approach on the test data. A correct implementation should achieve approximately 90% accuracy on test data.
7. Use `GridSearchCV` to tune the regularization parameter `C` (try values like 0.1, 1.0, 10.0) with 5-fold cross-validation.

#### 2.1 Normalization via Min-Max Normalization

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

#### 2.2.1 Process Labels

```python
from pandas import Series
import pandas as pd

def label_to_binary(desired_number: int, labels: Series):
    parsed_labels = []
    for label in labels:
        parsed_int = int(label)
        if parsed_int == desired_number:
            parsed_labels.append(1)
        else:
            parsed_labels.append(0)
    return parsed_labels

# one vs rest processing for y_train
y_train_9_list = label_to_binary(9, y_train)

#parse to pandas array
y_train_9 = pd.array(y_train_9_list)

# one vs rest processing for y_train
y_test_9_list = label_to_binary(9, y_test)

#parse to pandas array
y_test_9 = pd.array(y_test_9_list)

# print(list(y_train))
# print(y_train_9_list)
          
```

#### 2.2.2 Train Model with LogReg

```python
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(random_state=1, max_iter=1000)

# train model
logreg.fit(X_train_scaled, y_train_9)
```

#### 2.2.3 Calculate Accuracy

```python
from sklearn.metrics import accuracy_score

# predictions on test dataset
predictions = logreg.predict(X_test_scaled)

accuracy_score(y_test_9, predictions)
```

#### 2.3 Models for each digit

```python
models = []
for digit in np.arange(0, 10):
    y_train_digit_list = label_to_binary(digit, y_train)
    y_train_digit = pd.array(y_train_digit_list)
    
    logreg = LogisticRegression(random_state=1, max_iter=1000)
    logreg.fit(X_train_scaled, y_train_digit)
    
    models.append(logreg)
```

#### 2.4 Get optimal model for each datapoint

```python
test_predictions = []

for image in X_test_scaled:
    highest_prediction = -1
    best_model = -1

    for i, model in enumerate(models):
        prob = model.predict_proba([image])[0][1]
        if prob > highest_prediction:
            highest_prediction = prob
            best_model = i

    test_predictions.append(best_model)
```

#### 2.5 Calculate the accuracy

```python
accuracy_score(y_test, test_predictions)
```

#### 2.6 GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

param_grid = {"C": [0.1, 1.0, 10.0]}

grid = GridSearchCV(
    estimator=LogisticRegression(max_iter=1000),
    param_grid=param_grid,
    cv=5
)

grid.fit(X_train_scaled, y_train)
print(grid.best_params_)
print(grid.best_score_)
```
